# ***`array dimension`***

```python
# 1d
one=np.array([1,2,3])
print(one)
print(one.ndim,"\n")

#2d
two=np.array([
    [1,2,3],#row_1
    ["a","b","c"]#row_2
    ])
print(two)
print(two.ndim,"\n")

#3d
three=np.array([[1,2,3],["a","b","c"],[1,1.5,"a"]])
print(three)
print(three.ndim,"\n")#it showing 2 cause the last dim last data is a string  so numpy convert all the other data type into a string so it consider 2nd din and 3rd dim same

# 3d main: array that contains multiple 2d array
three_main=np.array([[[1,2,3],["a","b","c"]],
                     [[10,9,8],[2.5,6.5,8.8]]

                     ])
print(three_main)
print(three_main.ndim,"\n")
```

# ***`shape dtype and dim of array`***



```python
print(one.ndim,one.shape,one.dtype,"\n")#(shape will return(num of(col),num of(row))
print(two.ndim,two.shape,two.dtype,"\n")
print(three.ndim,three.shape,three.dtype,"\n")
print(three_main.ndim,three_main.shape,three_main.dtype)#(shape will return(num of (floor),num of(col),num of(row))
```

```python
arr_1=np.array([1,2,3])
print(arr_1.dtype)

arr_2=np.array([1,2,3.5])
print(arr_2.dtype)#upcasting

arr_3=np.array([1.1,2.5,3.5])
print(arr_3.dtype)
```

# ***`ND-Array Creation From Existing Data`***

```python
#suppose a list given
list_1=[10,20,20.5]
arr_1=np.array(list_1)
convert=arr_1.astype("int16")
print(arr_1)
print(type(arr_1))
print(arr_1.dtype)
print(convert)
print(convert.dtype)
```

# ***`Random value genrate`***

```python
arr=np.random.rand(2,3)
print(arr)
print(arr.dtype)
print(arr.size)
print(arr.shape)
print(arr.ndim)
print(type(arr))
```

# ***`Int Generate`***
