```python
import numpy as np

data=np.genfromtxt("./student_scores.csv",delimiter=",",skip_header=1)
print(data)
```

```python
Student_ID=data[:,0]
print(f"Student_ID:{Student_ID.T}\n")

Math_marks=data[:,1]
print(f"Math_marks:{Math_marks.T}\n")

Phy_marks=data[:,2]
print(f"Phy_marks:{Phy_marks.T}\n")

Chem_marks=data[:,3]
print(f"Chem_marks:{Chem_marks.T}\n")
```

```python
#finding max num of math phy chem

max_math=np.max(Math_marks)
max_phy=np.max(Phy_marks)
max_chem=np.max(Chem_marks)

print(f"max_math:{max_math}\n")
print(f"max_phy:{max_phy}\n")
print(f"max_chem:{max_chem}\n")
```
