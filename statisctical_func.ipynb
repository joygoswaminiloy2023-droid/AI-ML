```python
import numpy as np

data=np.genfromtxt("./student_scores.csv",delimiter=",",skip_header=1)
print(data)
```

```python
Student_ID=data[:,0]
print(f"Student_ID:{Student_ID.T}\n")

Math_marks=data[:,1]
print(f"Math_marks:{Math_marks.T}\n")

Phy_marks=data[:,2]
print(f"Phy_marks:{Phy_marks.T}\n")

Chem_marks=data[:,3]
print(f"Chem_marks:{Chem_marks.T}\n")
```

```python
#finding max,min,avg,median,standard devision num of math phy chem
print("Max")
max_math=np.max(Math_marks)
max_phy=np.max(Phy_marks)
max_chem=np.max(Chem_marks)

print(f"max_math:{max_math}\n")
print(f"max_phy:{max_phy}\n")
print(f"max_chem:{max_chem}\n\n")

#min
print("Min")
min_math=np.min(Math_marks)
min_phy=np.min(Phy_marks)
min_chem=np.min(Chem_marks)

print(f"min_math:{min_math}\n")
print(f"min_phy:{min_phy}\n")
print(f"min_chem:{min_chem}\n\n")

# avg

print("Avg")
avg_math=np.mean(Math_marks)
avg_phy=np.mean(Phy_marks)
avg_chem=np.mean(Chem_marks)

print(f"avg_math:{avg_math:.2f}\n")
print(f"avg_phy:{avg_phy:.2f}\n")
print(f"avg_chem:{avg_chem:.2f}\n\n")

#median
print("Median")
median_math=np.median(Math_marks)
median_phy=np.median(Phy_marks)
median_chem=np  .median(Chem_marks)

print(f"median_math:{median_math:.2f}\n")
print(f"median_phy:{median_phy:.2f}\n")
print(f"median_chem:{median_chem:.2f}\n\n")

#standard deviation

std_math=np.std(Math_marks)
std_phy=np.std(Phy_marks)
std_chem=np.std(Chem_marks)

print("standard deviation")
print(f"std_math:{std_math:.2f}\n")
print(f"std_phy:{std_phy:.2f}\n")
print(f"std_chem:{std_chem:.2f}\n\n")
```
