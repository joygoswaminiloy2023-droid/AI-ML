# ***`Filtering`***

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah","Sakib"],
    "City": ["New York", "Los Angeles", "Newark", "Boston", "New Delhi", "Chicago", "New Orleans", "Houston","H Los Ang"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "HR", "HR"],
    "Salary": [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000,70000]
}

df = pd.DataFrame(data)
```

```python
df
```

# ***`Finding String by Keyword`***

```python
# finding the string which has new word
df.loc[df['City'].str.contains('New')]
```

```python
#finding the city info of the persons who's country name contain "ton"

df.loc[df['City'].str.contains('ton')]
```

```python
#find only the names of person whos city name start with "Los"
df.loc[df['City'].str.contains('^los',case=False),['Name','Department']]
```

```python
#find only the names of person whos city name ends with "ang
df.loc[df['City'].str.contains('Ang$',case=False),["Name","Department"]]
```

```python
#Find the name and salary of all the persons whos  name start with vowel
df.loc[df['Name'].str.contains('^[aeiou]',case=False),["Name","Salary"]]
```

```python
#using or and condition
df.loc[df['City'].str.contains('New|Los')]
```
