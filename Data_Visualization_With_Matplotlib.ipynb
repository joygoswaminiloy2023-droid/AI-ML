# ***`2D Line Plot Basics`***

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
s1_Study_hours=[7,7,6,4,7,2,1]
s2_Study_hours=[6,5,4,2,7,7,2]

Week_days=[1,2,3,4,5,6,7]

plt.xlabel("Study_Hours->")
plt.ylabel("Week_Days->")
plt.title("Study Consistancy Graph")

plt.plot(Week_days,s1_Study_hours)
plt.plot(Week_days,s2_Study_hours)

plt.show()


```

```python
df=pd.read_csv('enrollment_data.csv')
df
```

# ***`2d visualization by pandas df`***

```python
plt.plot(df['Year'],df['Programming'])
plt.plot(df['Year'],df['Digital Marketing'])
plt.title('Enrollment Trend')
plt.xlabel('Year')
plt.ylabel('Enrollment & Digital Marketing')
plt.show()
```

```python
#showing the indicator or legend

plt.plot(df['Year'],df['Programming'],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')

plt.title('Enrollment Trend')
plt.xlabel('Year')
plt.ylabel('Enrollment & Digital Marketing')

plt.legend()# showing the indicator
plt.show()

```

```python
#legend location

plt.plot(df['Year'],df['Programming'],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')

plt.title('Enrollment Trend')
plt.xlabel('Year')
plt.ylabel('Enrollment & Digital Marketing')

plt.legend(loc='lower right')# showing the indicator in lower right

plt.show()
```

```python
#color,width,linestyle,marker,grid Changing of the line

plt.plot(df['Year'],df['Programming'],label='Programming',color="#42ecf5",linewidth=3,marker='o',markersize=7)
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing',color='red',linestyle='dashed')

plt.title('Enrollment Trend')
plt.xlabel('Year')
plt.ylabel('Enrollment & Digital Marketing')

plt.legend(loc='lower right')# showing the indicator in lower right
plt.grid()
plt.show()
```

# ***`Scatter Plots`***

```python
students=pd.read_csv('student_IQdata.csv')
students
```

```python
#relation between Student_IQ & Hours
plt.scatter(students['Study_Hour'], students['IQ_Score'])
plt.title('Student IQ vs Study Hours')
plt.xlabel('Study Hours')
plt.ylabel('IQ Score')
plt.show()
```

# ***`Histogram`***

```python
plt.hist(students['IQ_Score'],color='red',edgecolor='black')
plt.xlabel('IQ_Score')
plt.ylabel('Frequency')
plt.title('Histogram of IQ_Score')
plt.show()
```

```python
plt.hist(students['Study_Hour'],color='green',edgecolor='black')
plt.grid(axis='y')
plt.xlabel('Study_Hour')
plt.ylabel('Frequency')
plt.title('Histogram of Study_Hour & IQ_Score')
plt.show()
```

```python
#bins number
plt.hist(students['Study_Hour'],color='green',bins=25,edgecolor='black')
plt.grid(axis='y')
plt.xlabel('Study_Hour')
plt.ylabel('Frequency')
plt.title('Histogram of Study_Hour & IQ_Score')
plt.show()
```

```python
#bins like list
plt.hist(students['Study_Hour'],color='green',bins=[1,2,3,4,5,6,7],edgecolor='black')
plt.grid(axis='y')
plt.xlabel('Study_Hour')
plt.ylabel('Frequency')
plt.title('Histogram of Study_Hour & IQ_Score')
plt.show()
```
