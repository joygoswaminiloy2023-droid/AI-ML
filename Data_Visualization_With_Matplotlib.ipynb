# ***`2D Line Plot Basics`***

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
s1_Study_hours=[7,7,6,4,7,2,1]
s2_Study_hours=[6,5,4,2,7,7,2]

Week_days=[1,2,3,4,5,6,7]

plt.xlabel("Study_Hours->")
plt.ylabel("Week_Days->")
plt.title("Study Consistancy Graph")

plt.plot(Week_days,s1_Study_hours)
plt.plot(Week_days,s2_Study_hours)

plt.show()


```

```python
df=pd.read_csv('enrollment_data.csv')
df
```

# ***`2d visualization by pandas df`***

```python
plt.plot(df['Year'],df['Programming'])
plt.plot(df['Year'],df['Digital Marketing'])
plt.title('Enrollment Trend')
plt.xlabel('Year')
plt.ylabel('Enrollment & Digital Marketing')
plt.show()
```

```python
#showing the indicator or legend

plt.plot(df['Year'],df['Programming'],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')

plt.title('Enrollment Trend')
plt.xlabel('Year')
plt.ylabel('Enrollment & Digital Marketing')

plt.legend()# showing the indicator
plt.show()

```
