# Type-**Casting**

```python
# Get input from user (stored initially as strings)
a=input('Enter an int Num')
b=input('Enter an int Num')

# Print initial types to verify they are strings
print("a:",type(a),"b:",type(b))

# Convert (type-cast) both string values to floats
a=float(a)
b=float(b)

# Print types again to confirm conversion and output the sum
print("a:",type(a),"b:",type(b))
print(a+b)
```

```python
# Assign a floating-point number (10.6) to the variable 'num'
num=10.6

# Copy the reference/value of 'num' into a new variable 'num_1'
num_1=num

# Print the data type of both variables to show they are both floats
print(type(num),'/', type(num_1))
```

```python
row_of_num=input("Enter Row Of Numbers: ")
print("1.", row_of_num)
print("2.", type(row_of_num))

# so now we have to convert it into a list
print("****After Convert Into List****")
numbers=row_of_num.split()
print("3.", numbers)
print("4.", type(numbers))

# now grab the single values from the list
print("****After Grab Single Values From List****")
num_1=numbers[0]
num_2=numbers[1]
num_3=numbers[2]
print("5.", num_1)
print("6.", type(num_1))
print("7.", num_2)
print("8.", type(num_2))
print("9.", num_3)
print("10.", type(num_3))

# NOW TYPE_CAST FOR OPERATIONS
print("****After Type Casting****")
num_1=int(num_1)
num_2=int(num_2)
num_3=int(num_3)
print("11.",num_1,"|",type(num_1))
print("12.", num_2,"|", type(num_2))
print("13.", num_3,"|",type(num_3))



#lets Perform Operation
print("****After perform operation****")

print("14. SUM: ",num_1+num_2+num_3)

```

# ***Loop***

```python
# suppose i want to train a model for better accuracy so many times
#accuracy 90% is perfect for my model now i have to train the model and test accuricy untill 90% came every time accuracy increase 2%
accuracy=0
i=0
while accuracy<=90:

  i=i+1
  print(f"Model no:{i} | Training....current accuracy: {accuracy}%")
  accuracy=accuracy+2


print(f"Model No:{i} Perfect!")
```

# ***List***

```python
# suppose i have list of accuracies now i have to find the max accuracy for which k(here k is index)

accuracies=[.32,.35,.85,.98,.20,.12]
accuracy_input=input("Enter Accuracy: ")
accuracy_input=float(accuracy_input)

accuracies.append(accuracy_input)
print(accuracies)
max_acc=max(accuracies)
print(f"Best Accuracy: {max_acc}")
print("Value Of K For Best Accuracy:",accuracies.index(max_acc))
```
