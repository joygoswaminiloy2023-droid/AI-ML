# ***Array Manupulation***

## ***`Shape Changing`***




```python
import numpy as np

arr=np.random.randint(1,100,size=(10,5))
print(arr,"\n")
print(f"Shape:{arr.shape}")
b=arr.reshape(5,10)
print(b,"\n")
print(f"Shape:{b.shape}")

c=arr.reshape(10,3)#not possible cause arr containing 50 data but c can contain 30 data
print(c,"\n")
print(f"Shape:{c.shape}")
```

# ***`ravel dynamic solution`***

it can chnge dimension both column and row wise

```python
arr_2=np.random.randint(1,10,size=(5,5))
print(arr_2,"\n")

Row_WISE_FLATTENING=arr_2.ravel(order="c")
print(f"Row_Wise:{Row_WISE_FLATTENING}\n")

coloumn_WISE_FLATTENING=arr_2.ravel(order="f")
print(f"Column_Wise:{coloumn_WISE_FLATTENING}\n")
```

# ***`Concatenate`***

```python
arr=np.random.randint(1,10,size=(5,5))
print(arr,"\n")
arr_2=np.random.randint(11,20,size=(5,5))
print(arr_2,"\n")

con=np.concatenate((arr,arr_2),axis=0)#row Wise
print(f"Row_wise:\n{con}\n")

con_2=np.concatenate((arr,arr_2),axis=1)#column Wise
print(f"Col_wise:\n{con_2}\n")
```





*   **whenever concatenate row wise col must same**
*   **col wise row must same**





```python
arr=np.random.randint(1,10,size=(4,5))
print(arr,"\n")
arr_2=np.random.randint(11,20,size=(3,5))
print(arr_2,"\n")

row_wise=np.concatenate((arr,arr_2),axis=0)
print(f"Row_wise:\n{row_wise}\n")

col_wise=np.concatenate((arr,arr_2),axis=1)
print(f"Col_wise:\n{col_wise}\n")
```

# ***`2d Sorting`***

```python
import numpy as np
arr=np.array([[10,3,5],[8,9,5]])
#verticle:
ver_sort=np.sort(arr,axis=0)
print(f"Verticle_sort:\n{ver_sort}\n")

#horizon:
hor_sort=np.sort(arr,axis=1)
print(f"Horizon_sort:\n{hor_sort}")
```

# ***`Searching`***

```python
import numpy as np
#1d
x=np.array([10,8,11,5,3])
print(x)
search=np.where(x==8)
print(search,"\n")

#2d
print(arr)
search_from_mat=np.where(arr==5)
print(search_from_mat,"\n")

#1d
print(x,"\n")
replace_value=np.where(x==8,8,1)#it replace values with 1 except 8
print(replace_value,"\n")

replace_2=np.where(x>5,x,0)#it will replace values with 0 if the value less than or 5
print(replace_2,"\n")

#2d
print(arr)
index=np.where(arr==5)
print(index,"\n")

replace=np.where(arr==5,0,5)
print(replace,"\n")

```
