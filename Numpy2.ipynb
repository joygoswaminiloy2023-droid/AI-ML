# ***Array Manupulation***

## ***`Shape Changing`***




```python
import numpy as np

arr=np.random.randint(1,100,size=(10,5))
print(arr,"\n")
print(f"Shape:{arr.shape}")
b=arr.reshape(5,10)
print(b,"\n")
print(f"Shape:{b.shape}")

c=arr.reshape(10,3)#not possible cause arr containing 50 data but c can contain 30 data
print(c,"\n")
print(f"Shape:{c.shape}")
```

# ***`Dimension Manupulate`***

---



```python
print(arr,"\n")
print(f"Dimension:{arr.ndim}")
flatten=arr.flatten()
print(flatten,"\n")
print(f"Dimension:{flatten.ndim}")
```
