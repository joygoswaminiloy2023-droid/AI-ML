```python
import numpy as np
import pandas as pd
import matplotlib as plt
import seaborn as sns
```

```python
df=pd.read_csv('sns_data.csv')
df.head()
```

```python
#axis Level
sns.lineplot(data=df,x='week',y='attendance_rate',errorbar=None,hue="gender")
```

```python
#Figure LEvel
sns.relplot(kind='line',data=df,x='week',y='attendance_rate',errorbar=None,hue="gender")
```

# ***`Sactter Plot Using Seaborn`***

```python
sns.scatterplot(data=df,x='study_hours',y='test_score')
```

```python
sns.scatterplot(data=df,x='study_hours',y='test_score',hue='gender',style='class_level',size="Tshirt_size")
```

```python
#figure Level
sns.relplot(kind='scatter',data=df,x='study_hours',y='test_score',hue='gender',style='class_level',size="Tshirt_size")
```

```python
load_data=sns.load_dataset('tips')
load_data.head()
```

```python
sns.scatterplot(data=load_data,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
sns.relplot(kind='scatter',data=load_data,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

# ***`Facet Plot Using Seaborn`***

```python
student_data=pd.read_csv('student_dataset_complete.csv')
student_data.head()
```

```python
sns.scatterplot(data=student_data,x='study_hours',y='test_score',hue='gender')
```

```python
sns.relplot(kind='scatter',data=student_data,x='study_hours',y='test_score',hue='gender',col='gender',row='hostel')
```

# ***`Histogram`***

```python
sns.histplot(data=student_data,x='attendance_rate',hue='gender',bins=15,element='step')
```

```python
sns.displot(kind='hist',data=student_data,x='attendance_rate',col='gender',row='study_hour')
```

# ***`KDE using HistoGram`***

```python
sns.kdeplot(data=student_data,x='study_hours')
```

```python
sns.displot(kind='kde',data=student_data,x='study_hours',hue='gender')
```

# ***`Count Plot and Bar Plot using Seaborn`***

```python
sns.countplot(data=student_data,x='gender')
```

```python
sns.countplot(data=student_data,x='subject',hue='gender')
```

Regression Plot

```python
sns.regplot(data=student_data,x='study_hours',y='test_score')
```

```python
sns.lmplot(data=student_data,x='study_hours',y='test_score',hue='gender')
```

# ***`Pair Plots`***

```python
students_marks=student_data[['Marks','study_hours','attendance_rate','gender']]
students_marks
```

```python
sns.pairplot(students_marks)
```

```python
sns.pairplot(students_marks,kind='hist')
```

```python
sns.pairplot(students_marks,hue='gender')
```
