# ***Ans To The Qus No.`02`***

```python
def compute_ratio(x,y):
  if (x*y)==0:
    print('Error: not divisable By Zero')

  else:
    process=(x+y)/(x*y)
    return process

x=int(input())
y=int(input())
compute_ratio(x,y)
```

# ***Ans To The Qus No.`05`***

```python
def reverse(word):
  rev=''.join(reversed(word))
  if rev==word:
    print('Palindrome')
  else:
    print('Not Palindrome')

word=input().upper()
reverse(word)
```
