# ***`Diffrent Files Loading With Pandas`***

```python
#CSV_File Load
csv_data=pd.read_csv("student_data.csv")
print(csv_data)
print(type(csv_data),"\n\n")
```

```python
#Xlsx(Excel) File Load
excel_data=pd.read_excel("phitron_student_marks.xlsx")
print(excel_data)
print(type(excel_data),"\n\n")
```

```python
#parquet file
par_data=pd.read_parquet("students.parquet")
print(par_data)
print(type(par_data),"\n\n")
```

```python
#json

json_data=pd.read_json("data.json")
print(json_data)
print(type(json_data),"\n\n")
```
