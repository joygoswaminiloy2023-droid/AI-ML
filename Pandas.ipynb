# ***`Dataframe from Inbuilt Data with Column and Indexes`***

```python
#list to dataframe
my_list=[['Joy',69],['Nil',64]]
print(my_list,"\n\n")
list_df=pd.DataFrame(my_list)
print(list_df)
```

```python
#mention the column name and index according to my preference
list_df=pd.DataFrame(my_list,columns=['Name','ID'],index=[1,2])
print(list_df)
```

```python
#tuple to dataframe
my_tuple=(("Joy",69),("Nil",64))
tup_df=pd.DataFrame(my_tuple, columns=["Name","ID"],index=[1,2])
tup_df
```

```python
#dictionary to data frame

my_dic=[
    {'Name':'Alice','Age':'25','Gender':'Female'},
    {'Name':'Bob','Age':'30','Gender':'Male'},
    {'Name':'Charlie','Age':'35','Gender':'Male'},
    {'Name':'David','Age':'40'}

]

dic_df=pd.DataFrame(my_dic)
dic_df
```

# ***`Index_Changing`***

```python
old=df.loc[3:7,["StudentID","Data Structure Marks","Algorithm Marks"]]

df_ind=df.set_index('StudentID')

new=df_ind.iloc[3:7,1:3]#using iloc cause now the index is not numeric

print(f"OLD:\n{old}\n")
print(f"NEW:\n\n{new}\n")
```

```python
#rename(column={"previous_col_name":"new"})
df_ind.rename(columns={
    "FullName": "FN",
    "Python Marks": "P_Marks"
},inplace=True)
df_ind
```
